## Feature Engineering

```python
# Copia independente do df_clean
mandantes = df_clean[[
    "ano_campeonato",
    "data",
    "rodada",
    "time_mandante",
    "time_visitante",
    "gols_mandante",
    "gols_visitante"
]].copy()

# Renomeando a coluna time_mandade e visitante para time e adversário e adiciona 1 em todas as linhas
mandantes = mandantes.rename(columns={
    "time_mandante": "time",
    "time_visitante": "adversario",
    "gols_mandante": "gols_marcados",
    "gols_visitante": "gols_sofridos"
})

mandantes["casa"] = 1

mandantes
```

```python
# Copia todos os dados indepententes do df_clean
visitantes = df_clean[[
    "ano_campeonato",
    "data",
    "rodada",
    "time_mandante",
    "time_visitante",
    "gols_mandante",
    "gols_visitante"
]].copy()

#renomeia as colunas time visitante para time e mandante para adversário e adiciona 0 em todas as linhas
visitantes = visitantes.rename(columns={
    "time_visitante": "time",
    "time_mandante": "adversario",
    "gols_visitante": "gols_marcados",
    "gols_mandante": "gols_sofridos"
})

visitantes["casa"] = 0
visitantes
```

Vamos usar o Df_times e o DF clean para o modelo. Ambos tem funções diferentes no projeto. df clean existe porque ele representa o campeonato no nível de partida (1 linha = 1 jogo), já o df_times representa o campeonato no nível de time (1 jogo= 2 linhas)
Ambos irão se unir no Dataframe final, df_modelo.

```python
# Junta os Dois DataFrames (um embaixo do outro) - repare que duplicou o tamanho do df, porém agora ele mostra quem é o time e onde esta jogando (em casa ou fora)
# As duplicadas não representam um erro no dataset, pois cada linha virou uma observação para cada time (visitante e mandante) => Cada linha representa uma coisa diferente!
df_times = pd.concat(
    [mandantes, visitantes],
    ignore_index=True
)

df_times
```

```python
# Verificando se existem algum jogo duplicado - Desencargo
df_times.duplicated().sum()
```

```python
# Ordenar
df_times = df_times.sort_values(
    ["time", "data"]
).reset_index(drop=True)
```

```python
#Criação dos Pontos de acordo com o resultado da partida (3 pontos para vitória, 1 ponto para empate e 0 pontos para derrota)

def calcular_pontos(row):
    if row["gols_marcados"] > row["gols_sofridos"]:
        return 3
    elif row["gols_marcados"] == row["gols_sofridos"]:
        return 1
    else:
        return 0

df_times["pontos"] = df_times.apply(  #aplly() serve para aplicar uma função em um dataframe
    calcular_pontos,
    axis=1 # Aplica a função linha por linha, cada linha é passada para calcular_pontos como uma series
           #O valor retornado pela função vira o valor da nova coluna "pontos" naquela linha
)
```

```python
#Calculo do Saldo de Gols
df_times["saldo_gols"] = (
    df_times["gols_marcados"]
    - df_times["gols_sofridos"]
)
```

```python
df_times
```

```python
# Ordena o DataFrame, seguindo a ordem: TIme, ano_campeonato, Data. Além de resetar os indices (0,1,2,3)
df_times = df_times.sort_values(
    ["time", "ano_campeonato", "data", "rodada"]
).reset_index(drop=True)
```

```python
# Cria a coluna que valida quem ganhou a partida da rodada - quem fez mais gols venceu - Empate? Esta sendo considerado como derrota (estão juntos na mesma classificação) - corrigir?
df_times["vitoria"] = (
    df_times["gols_marcados"] > df_times["gols_sofridos"]
).astype(int)
```

```python
# Para cada time + ano_campeonato, pegue a coluna gols marcados e chame ela de X
# O groupby faz com que o lambda faça as contas separadamente para cada time dentro de cada ano (Flamengo 2022 - pega todas as linhas do flamengo de 2022)
# O shift(1) desloca 1 posição para baixo para pegar os jogos das rodadas anteriores
# O rollings cria uma janela olhando para os últimos 5 valores
df_times["media_gols_5"] = (
    df_times
    .groupby(["time", "ano_campeonato"])["gols_marcados"]
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=5).mean()
    )
)
```

```python
# Mesmo esquema (praticamente) da media de gols dos ultimos 5 jogos
df_times["media_gols_sofridos_5"] = (
    df_times
    .groupby(["time", "ano_campeonato"])["gols_sofridos"]
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=5).mean()
    )
)
```

```python
# Mesmo esquema (praticamente) da media de gols dos ultimos 5 jogos
df_times["pontos_ultimos_5"] = (
    df_times
    .groupby(["time", "ano_campeonato"])["pontos"]
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=5).sum()
    )
)
```

```python
# Mesmo esquema (praticamente) da media de gols dos ultimos 5 jogos, unica diferença é que agora o groupby vai atribuir as qtd de vitorias no lugar dos gols
df_times["vitorias_ultimos_5"] = (
    df_times
    .groupby(["time", "ano_campeonato"])["vitoria"]
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=5).sum()
    )
)
```

```python
# Mesmo esquema (praticamente) da media de gols dos ultimos 5 jogos, unica diferença é que agora o groupby vai atribuir a qtd de pontos.
df_times["jogos_na_janela_5"] = (
    df_times
    .groupby(["time", "ano_campeonato"])["pontos"]
    .transform(
        lambda x: x.shift(1).rolling(5, min_periods=5).count()
    )
)
# Valida o aproveitamento dos ultimos 5 jogos - Vai te 0 a 1. Multiplica por 3 os jogos para ficar na mesma escala q a pontuação (vitoria = 3 pontos)
df_times["aproveitamento_5"] = (
    df_times["pontos_ultimos_5"]
    / (df_times["jogos_na_janela_5"] * 3)
)
```

```python
# Elo -  todo mundo começa com 1500 
# Antes de cada jogo, calcula a probabilidade esperada de cada time vencer com base na diferença de Elo
# Depois do resultado, atualiza o Elo
# Deve considerar apenas o Elo antes da partida, não o depois.


#Validar cada linha do Df_clean - so tem 1 linha por rodada

def valida_elo(df,K=18, ano_inicio_erro = None, ano_fim_erro = None): #provisório 8 => tem o menor erro até agora (teste abaixo)
    elos= {}
    erros= []
    for i, row in df.iterrows():

        mandante = row['time_mandante']
        visitante = row['time_visitante']

        if mandante not in elos:
            elos[mandante] = 1500
        if visitante not in elos:    
            elos[visitante] = 1500

        R_A = elos[mandante]
        R_B = elos[visitante]

        # Salva o Elo antes da partida atual( rodada a ser calculada)
        df.loc[i,'elo_mandante_pre_jogo'] = R_A
        df.loc[i,'elo_visitante_pre_jogo'] = R_B 

        E_A = 1 / (1 + 10**((R_B - R_A)/400)) # Expectativa de vitoria time A de 0 a 1
        E_B = 1 / (1 + 10**((R_A - R_B)/400)) # Expectativa de vitoria Time B

        # Parâmetro que controla o quanto a pontuação Elo pode variar depois de cada partida

        if row['gols_mandante'] > row['gols_visitante']:
            S_A = 1
            S_B = 0
            
        elif row['gols_mandante'] == row['gols_visitante']:
            S_A = 0.5
            S_B = 0.5
        else:
            S_A = 0 
            S_B = 1

        R_novo_A = R_A + K * (S_A - E_A)
        R_novo_B = R_B + K * (S_B - E_B)
        erro_A = (S_A - E_A) ** 2
        erro_B = (S_B - E_B) ** 2

        ano = row['ano_campeonato']
        if ano_inicio_erro is not None and ano_fim_erro is not None:
            if ano_inicio_erro<=ano<=ano_fim_erro:
                erros.append((erro_A + erro_B) / 2)

        #erros.append((erro_A + erro_B) / 2)

        elos[mandante] = R_novo_A
        elos[visitante] = R_novo_B

    if ano_inicio_erro is not None and ano_fim_erro is not None:
        return df, sum(erros)/len(erros)
    else:
        return df

#Garantir a ordem cronologica do Df
df_clean = df_clean.sort_values(['data', 'rodada']).reset_index(drop=True)

df_clean = valida_elo(df_clean) # a fução retorna uma tupla, sendo o dataframe no incdice 0 e o erro no indice 1.


for K in [10, 12, 14, 16, 18, 20, 25, 30]:
    _, erro = valida_elo(
        df_clean.copy(),
        K=K,
        ano_inicio_erro=2019,
        ano_fim_erro=2021
    )
    print(K, erro)
# USAMOS 2019 A 2021 para validar o 2022 a 2024 e 18 tem o menor erro, vamos seguir com ele por enquanto, pois ele n explodiu ao testar em 2022 a 2024.
```

```python
def cria_df_modelo (df_times,df_clean):


    for i, row in df_times.iterrows():

        mask_1 = (
        (df_clean["data"] == row["data"]) & #Funciona como um 'E', mas no pandas ele compara posição por posição, como se fosse um and, mas para as séries inteiras do pandas (valida toda as sries da condição e retorna as linhas q tem todas as colunas/series com true)
        (df_clean["rodada"] == row["rodada"])& #N usamos o 'AND', pois ele é usado normalmente entre um booleano e outro booleano e , no pandas, temos séries inteiras de booleanos
        (df_clean['time_mandante'] == row['time'])&
        (df_clean["time_visitante"] == row["adversario"])&
        (df_clean["ano_campeonato"] == row["ano_campeonato"])
    )

        mask_0 = (
            (df_clean["data"] == row["data"]) &
            (df_clean["rodada"] == row["rodada"])&
            (df_clean['time_mandante'] == row['adversario'])&
            (df_clean["time_visitante"] == row["time"])&
            (df_clean["ano_campeonato"] == row["ano_campeonato"])
        )


        if row["casa"] == 1:

            df_clean.loc[mask_1, "media_gols_5_mandante"] = row["media_gols_5"]
            df_clean.loc[mask_1, "media_gols_sofridos_5_mandante"] = row["media_gols_sofridos_5"]
            df_clean.loc[mask_1, "pontos_ultimos_5_mandante"] = row["pontos_ultimos_5"]
            df_clean.loc[mask_1, "vitorias_ultimos_5_mandante"] = row["vitorias_ultimos_5"]

            df_clean.loc[mask_1, "media_gols_mesmo_mando_5_mandante"] = row["media_gols_mesmo_mando_5"]
            df_clean.loc[mask_1, "media_gols_sofridos_mesmo_mando_5_mandante"] = row["media_gols_sofridos_mesmo_mando_5"]
            df_clean.loc[mask_1, "pontos_mesmo_mando_ultimos_5_mandante"] = row["pontos_mesmo_mando_ultimos_5"]
            df_clean.loc[mask_1, "aproveitamento_mesmo_mando_5_mandante"] = row["aproveitamento_mesmo_mando_5"]
            df_clean.loc[mask_1, "vitorias_mesmo_mando_ultimos_5_mandante"] = row["vitorias_mesmo_mando_ultimos_5"]

        else:

            df_clean.loc[mask_0, "media_gols_5_visitante"] = row["media_gols_5"]
            df_clean.loc[mask_0, "media_gols_sofridos_5_visitante"] = row["media_gols_sofridos_5"]
            df_clean.loc[mask_0, "pontos_ultimos_5_visitante"] = row["pontos_ultimos_5"]
            df_clean.loc[mask_0, "vitorias_ultimos_5_visitante"] = row["vitorias_ultimos_5"]

            df_clean.loc[mask_0, "media_gols_mesmo_mando_5_visitante"] = row["media_gols_mesmo_mando_5"]
            df_clean.loc[mask_0, "media_gols_sofridos_mesmo_mando_5_visitante"] = row["media_gols_sofridos_mesmo_mando_5"]
            df_clean.loc[mask_0, "pontos_mesmo_mando_ultimos_5_visitante"] = row["pontos_mesmo_mando_ultimos_5"]
            df_clean.loc[mask_0, "aproveitamento_mesmo_mando_5_visitante"] = row["aproveitamento_mesmo_mando_5"]
            df_clean.loc[mask_0, "vitorias_mesmo_mando_ultimos_5_visitante"] = row["vitorias_mesmo_mando_ultimos_5"]
    return df_clean
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss

def valida_logistica(C=1.0):

    resultados = []

    for ano_val in anos_validacao:

        train_mask = (
            (df_modelo["ano_campeonato"] >= 2012) &
            (df_modelo["ano_campeonato"] < ano_val)
        )

        val_mask = (
            df_modelo["ano_campeonato"] == ano_val
        )

        treino_X = X.loc[train_mask]
        treino_y = y.loc[train_mask]

        val_X = X.loc[val_mask]
        val_y = y.loc[val_mask]

        # Scaler aprende APENAS com o treino daquele fold
        scaler_cv = StandardScaler()

        treino_X_scaled = scaler_cv.fit_transform(treino_X)
        val_X_scaled = scaler_cv.transform(val_X)

        modelo_cv = LogisticRegression(
            C=C,
            max_iter=1000
        )

        modelo_cv.fit(
            treino_X_scaled,
            treino_y
        )

        proba_cv = modelo_cv.predict_proba(
            val_X_scaled
        )

        accuracy_cv = modelo_cv.score(
            val_X_scaled,
            val_y
        )

        logloss_cv = log_loss(
            val_y,
            proba_cv,
            labels=modelo_cv.classes_
        )

        resultados.append({
            "ano_validacao": ano_val,
            "accuracy": accuracy_cv,
            "log_loss": logloss_cv
        })

    return pd.DataFrame(resultados)
```

## Modelo A com pesos temporais

- Como o futebol varia muito de acordo com os anos, iremos usar uma meia vida de 2 anos, ou seja, a cada 2 anos o peso diminui.


- linha recente       → peso 1.00 ; 
2 anos atrás        → peso 0.50 ; 
4 anos atrás        → peso 0.25 ; 
6 anos atrás        → peso 0.125 ; 

# Catboost com pesos temporais

```python
def valida_catboosting_pesos(depth, learning_rate, iterations, l2_leaf_reg):

    resultados = []

    for ano_val in anos_validacao:

        train_mask = (
            (df_modelo["ano_campeonato"] >= 2012) &
            (df_modelo["ano_campeonato"] < ano_val)
        )

        val_mask = (
            df_modelo["ano_campeonato"] == ano_val
        )

        treino_X = X.loc[train_mask]
        treino_y = y_xgb.loc[train_mask]

        val_X = X.loc[val_mask]
        val_y = y_xgb.loc[val_mask]
        

        # Pega a maior data (mais recente) dos dados separados para treino
        data_ref = df_modelo.loc[treino_X.index, "data"].max()

        #vai pegar a diferença da data mais recente da data de determinada linha e dividir por 365.25 (considerando ano bissextos) para então saber quantos anos está de distancia dessa data.
        idade_anos = (
            data_ref - df_modelo.loc[treino_X.index,"data"]
        ).dt.days / 365.25

        # Se a diferença de anos for maior q 2 , o peso diminui em 0.5 (2/2 = 1)
        pesos_treino = 0.5 ** (idade_anos/3)

        modelo_cv = CatBoostClassifier(
            loss_function="MultiClass",
            depth=depth,
            learning_rate=learning_rate,
            iterations=iterations,
            l2_leaf_reg=l2_leaf_reg,
            random_seed=42,
            verbose=False
        )

        modelo_cv.fit(treino_X, treino_y,
            sample_weight = pesos_treino)

        proba_cv = modelo_cv.predict_proba(val_X)

        accuracy_cv = modelo_cv.score(
            val_X,
            val_y
        )

        logloss_cv = log_loss(
            val_y,
            proba_cv,
            labels=[0, 1, 2]
        )

        resultados.append({
            "ano_validacao": ano_val,
            "accuracy": accuracy_cv,
            "log_loss": logloss_cv
        })

    resultados_df = pd.DataFrame(resultados)

    return resultados_df
```

```python
def cria_elo_sazonal(df, K=18, alpha=0.75):

    df = df.sort_values(["data", "rodada"]).copy()

    elos = {}
    ano_anterior = None

    for i, row in df.iterrows():

        ano_atual = row["ano_campeonato"]

        # Quando muda a temporada, aproxima todos os Elos de 1500
        if ano_anterior is not None and ano_atual != ano_anterior:

            for time in elos:
                elos[time] = 1500 + alpha * (elos[time] - 1500)

        mandante = row["time_mandante"]
        visitante = row["time_visitante"]

        if mandante not in elos:
            elos[mandante] = 1500

        if visitante not in elos:
            elos[visitante] = 1500

        elo_mandante = elos[mandante]
        elo_visitante = elos[visitante]

        # Elo ANTES do jogo
        df.loc[i, "elo_mandante_sazonal"] = elo_mandante
        df.loc[i, "elo_visitante_sazonal"] = elo_visitante

        # Expectativa
        E_mandante = 1 / (
            1 + 10 ** ((elo_visitante - elo_mandante) / 400)
        )

        E_visitante = 1 - E_mandante

        # Resultado real
        if row["gols_mandante"] > row["gols_visitante"]:
            S_mandante = 1
            S_visitante = 0

        elif row["gols_mandante"] == row["gols_visitante"]:
            S_mandante = 0.5
            S_visitante = 0.5

        else:
            S_mandante = 0
            S_visitante = 1

        # Atualiza depois da partida
        elos[mandante] = (
            elo_mandante +
            K * (S_mandante - E_mandante)
        )

        elos[visitante] = (
            elo_visitante +
            K * (S_visitante - E_visitante)
        )

        ano_anterior = ano_atual

    return df
```

```python
#Função igual a catboost co pesos
def valida_catboost_B(
    depth=3,
    learning_rate=0.05,
    iterations=300,
    l2_leaf_reg=5,
    meia_vida=3
):

    resultados = []

    for ano_val in anos_validacao:

        train_mask = (
            (df_modelo_B["ano_campeonato"] >= 2012) &
            (df_modelo_B["ano_campeonato"] < ano_val)
        )

        val_mask = (
            df_modelo_B["ano_campeonato"] == ano_val
        )

        treino_X = X_B.loc[train_mask]
        treino_y = y_B_xgb.loc[train_mask]

        val_X = X_B.loc[val_mask]
        val_y = y_B_xgb.loc[val_mask]

        # Pesos temporais
        datas_treino = df_modelo_B.loc[treino_X.index, "data"]
        data_ref = datas_treino.max()

        idade_anos = (
            data_ref - datas_treino
        ).dt.days / 365.25

        pesos_treino = 0.5 ** (idade_anos / meia_vida)

        modelo = CatBoostClassifier(
            loss_function="MultiClass",
            depth=depth,
            learning_rate=learning_rate,
            iterations=iterations,
            l2_leaf_reg=l2_leaf_reg,
            random_seed=42,
            verbose=False
        )

        modelo.fit(
            treino_X,
            treino_y
        )

        proba = modelo.predict_proba(val_X)

        accuracy = modelo.score(val_X, val_y)

        logloss = log_loss(
            val_y,
            proba,
            labels=[0, 1, 2]
        )

        resultados.append({
            "ano_validacao": ano_val,
            "accuracy": accuracy,
            "log_loss": logloss
        })

    return pd.DataFrame(resultados)
```

## Regressa Logistica Modelo B

```python
from sklearn.calibration import calibration_curve
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss
import numpy as np
import pandas as pd


def valida_logistica_B(C=0.01):

    resultados = []
    resultados_calibracao = []

    for ano_val in anos_validacao:

        # -------------------------
        # 1. Separação temporal
        # -------------------------

        train_mask = (
            (df_modelo_B["ano_campeonato"] >= 2012) &
            (df_modelo_B["ano_campeonato"] < ano_val)
        )

        val_mask = (
            df_modelo_B["ano_campeonato"] == ano_val
        )

        treino_X = X_B.loc[train_mask]
        treino_y = y_B.loc[train_mask]

        val_X = X_B.loc[val_mask]
        val_y = y_B.loc[val_mask]

        # -------------------------
        # 2. Preencher NaN
        # -------------------------

        imputer = SimpleImputer(strategy="median")

        treino_X_imp = imputer.fit_transform(treino_X)
        val_X_imp = imputer.transform(val_X)

        # -------------------------
        # 3. StandardScaler
        # -------------------------

        scaler = StandardScaler()

        treino_X_scaled = scaler.fit_transform(treino_X_imp)
        val_X_scaled = scaler.transform(val_X_imp)

        # -------------------------
        # 4. Treinar modelo
        # -------------------------

        modelo = LogisticRegression(
            C=C,
            max_iter=1000
        )

        modelo.fit(
            treino_X_scaled,
            treino_y
        )

        # -------------------------
        # 5. Probabilidades
        # -------------------------

        proba = modelo.predict_proba(val_X_scaled)

        # -------------------------
        # 6. Accuracy
        # -------------------------

        accuracy = modelo.score(
            val_X_scaled,
            val_y
        )

        # -------------------------
        # 7. Log Loss
        # -------------------------

        logloss = log_loss(
            val_y,
            proba,
            labels=modelo.classes_
        )

        # -------------------------
        # 8. Brier multiclasses
        # -------------------------

        y_bin = label_binarize(
            val_y,
            classes=modelo.classes_
        )

        brier = np.mean(
            np.sum(
                (proba - y_bin) ** 2,
                axis=1
            )
        )

        resultados.append({
            "ano_validacao": ano_val,
            "accuracy": accuracy,
            "log_loss": logloss,
            "brier": brier
        })

        # -------------------------
        # 9. Calibração por classe
        # -------------------------

        for i, classe in enumerate(modelo.classes_):

            y_bin_classe = (
                val_y == classe
            ).astype(int)

            prob_true, prob_pred = calibration_curve(
                y_bin_classe,
                proba[:, i],
                n_bins=5,
                strategy="quantile"
            )

            for previsto, real in zip(
                prob_pred,
                prob_true
            ):

                resultados_calibracao.append({
                    "ano_validacao": ano_val,
                    "classe": classe,
                    "prob_prevista": previsto,
                    "freq_real": real
                })

    df_metricas = pd.DataFrame(resultados)

    df_calibracao = pd.DataFrame(
        resultados_calibracao
    )

    return df_metricas, df_calibracao
```

```python
from sklearn.pipeline import Pipeline
from sklearn.calibration import CalibratedClassifierCV
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import accuracy_score, log_loss
from sklearn.preprocessing import label_binarize
import numpy as np
import pandas as pd


def valida_logistica_B_calibrada(C=0.01, metodo="sigmoid"):

    resultados = []

    for ano_val in anos_validacao:

        # treino: tudo antes do ano de validação
        train_mask = (
            (df_modelo_B["ano_campeonato"] >= 2012) &
            (df_modelo_B["ano_campeonato"] < ano_val)
        )

        # validação: ano atual
        val_mask = (
            df_modelo_B["ano_campeonato"] == ano_val
        )

        # garante ordem temporal
        idx_treino = (
            df_modelo_B.loc[train_mask]
            .sort_values(["data", "rodada"])
            .index
        )

        idx_val = (
            df_modelo_B.loc[val_mask]
            .sort_values(["data", "rodada"])
            .index
        )

        treino_X = X_B.loc[idx_treino]
        treino_y = y_B.loc[idx_treino]

        val_X = X_B.loc[idx_val]
        val_y = y_B.loc[idx_val]

        if "target" in X_B.columns:
            print('target esta dentro')

        # pipeline da regressão original
        modelo_base = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            ),
            (
                "logistica",
                LogisticRegression(
                    C=C,
                    max_iter=1000
                )
            )
        ])

        # calibração interna respeitando tempo
        cv_temporal = TimeSeriesSplit(n_splits=3)

        modelo_calibrado = CalibratedClassifierCV(
            estimator=modelo_base,
            method=metodo,
            cv=cv_temporal
        )

        modelo_calibrado.fit(
            treino_X,
            treino_y
        )

        proba = modelo_calibrado.predict_proba(val_X)
        previsao = modelo_calibrado.predict(val_X)

        accuracy = accuracy_score(
            val_y,
            previsao
        )

        logloss = log_loss(
            val_y,
            proba,
            labels=modelo_calibrado.classes_
        )

        y_bin = label_binarize(
            val_y,
            classes=modelo_calibrado.classes_
        )

        brier = np.mean(
            np.sum(
                (proba - y_bin) ** 2,
                axis=1
            )
        )

        resultados.append({
            "ano_validacao": ano_val,
            "accuracy": accuracy,
            "log_loss": logloss,
            "brier": brier
        })

    return pd.DataFrame(resultados)
```

```python
def cria_df_times(df_jogos):

    mandante = pd.DataFrame({
        "ano_campeonato": df_jogos["ano_campeonato"],
        "data": df_jogos["data"],
        "rodada": df_jogos["rodada"],

        "time": df_jogos["time_mandante"],
        "adversario": df_jogos["time_visitante"],

        "casa": 1,

        "gols_marcados": df_jogos["gols_mandante"],
        "gols_sofridos": df_jogos["gols_visitante"]
    })

    visitante = pd.DataFrame({
        "ano_campeonato": df_jogos["ano_campeonato"],
        "data": df_jogos["data"],
        "rodada": df_jogos["rodada"],

        "time": df_jogos["time_visitante"],
        "adversario": df_jogos["time_mandante"],

        "casa": 0,

        "gols_marcados": df_jogos["gols_visitante"],
        "gols_sofridos": df_jogos["gols_mandante"]
    })

    df_times = pd.concat(
        [mandante, visitante],
        ignore_index=True
    )

    df_times["pontos"] = np.select(
        [
            df_times["gols_marcados"] > df_times["gols_sofridos"],
            df_times["gols_marcados"] == df_times["gols_sofridos"]
        ],
        [
            3,
            1
        ],
        default=0
    )

    df_times["saldo_gols"] = (
        df_times["gols_marcados"]
        - df_times["gols_sofridos"]
    )

    df_times["vitoria"] = (
        df_times["gols_marcados"]
        > df_times["gols_sofridos"]
    ).astype(int)

    df_times = df_times.sort_values(
        ["time", "data", "rodada"]
    ).reset_index(drop=True)

    return df_times
```

```python
def cria_df_modelo_B(df_times, df_jogos):

    df_modelo = df_jogos.copy()

    colunas_ewm = [
        "pontos_ewm",
        "gols_ewm",
        "gols_sofridos_ewm",
        "vitoria_ewm",
        "pontos_ewm_mando",
        "gols_ewm_mando",
        "gols_sofridos_ewm_mando",
        "vitoria_ewm_mando"
    ]

    # dados do mandante
    mandante = df_times[
        df_times["casa"] == 1
    ][
        [
            "ano_campeonato",
            "data",
            "rodada",
            "time"
        ] + colunas_ewm
    ].copy()

    mandante = mandante.rename(
        columns={
            "time": "time_mandante",
            **{
                col: f"{col}_mandante"
                for col in colunas_ewm
            }
        }
    )

    # dados do visitante
    visitante = df_times[
        df_times["casa"] == 0
    ][
        [
            "ano_campeonato",
            "data",
            "rodada",
            "time"
        ] + colunas_ewm
    ].copy()

    visitante = visitante.rename(
        columns={
            "time": "time_visitante",
            **{
                col: f"{col}_visitante"
                for col in colunas_ewm
            }
        }
    )

    # junta mandante
    df_modelo = df_modelo.merge(
        mandante,
        on=[
            "ano_campeonato",
            "data",
            "rodada",
            "time_mandante"
        ],
        how="left"
    )

    # junta visitante
    df_modelo = df_modelo.merge(
        visitante,
        on=[
            "ano_campeonato",
            "data",
            "rodada",
            "time_visitante"
        ],
        how="left"
    )

    return df_modelo
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, log_loss
from sklearn.preprocessing import label_binarize
import numpy as np
import pandas as pd

anos_teste = [2024, 2025, 2026]

resultados = []

for ano_teste in anos_teste:

    treino_mask = (
        (df_modelo_atual["ano_campeonato"] >= 2012) &
        (df_modelo_atual["ano_campeonato"] < ano_teste)
    )

    teste_mask = (
        df_modelo_atual["ano_campeonato"] == ano_teste
    )

    X_treino = df_modelo_atual.loc[
        treino_mask,
        features_modelo_B
    ]

    y_treino = df_modelo_atual.loc[
        treino_mask,
        "target"
    ]

    X_teste = df_modelo_atual.loc[
        teste_mask,
        features_modelo_B
    ]

    y_teste = df_modelo_atual.loc[
        teste_mask,
        "target"
    ]

    modelo = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "logistica",
            LogisticRegression(
                C=0.01,
                max_iter=1000
            )
        )
    ])

    modelo.fit(
        X_treino,
        y_treino
    )

    proba = modelo.predict_proba(X_teste)
    pred = modelo.predict(X_teste)

    accuracy = accuracy_score(
        y_teste,
        pred
    )

    ll = log_loss(
        y_teste,
        proba,
        labels=modelo.named_steps["logistica"].classes_
    )

    y_bin = label_binarize(
        y_teste,
        classes=modelo.named_steps["logistica"].classes_
    )

    brier = np.mean(
        np.sum(
            (proba - y_bin) ** 2,
            axis=1
        )
    )

    resultados.append({
        "ano_teste": ano_teste,
        "jogos": len(y_teste),
        "accuracy": accuracy,
        "log_loss": ll,
        "brier": brier
    })

validacao_recente = pd.DataFrame(resultados)

validacao_recente
```

```python
def calcula_elo_atual(df, K=18, alpha=0.75):
    jogos = df.sort_values(["data", "rodada"]).copy()

    elos = {}
    ano_anterior = None

    for _, row in jogos.iterrows():
        ano_atual = row["ano_campeonato"]

        # regressão à média quando muda a temporada
        if ano_anterior is not None and ano_atual != ano_anterior:
            for time in elos:
                elos[time] = 1500 + alpha * (elos[time] - 1500)

        mandante = row["time_mandante"]
        visitante = row["time_visitante"]

        if mandante not in elos:
            elos[mandante] = 1500

        if visitante not in elos:
            elos[visitante] = 1500

        elo_m = elos[mandante]
        elo_v = elos[visitante]

        esperado_m = 1 / (1 + 10 ** ((elo_v - elo_m) / 400))
        esperado_v = 1 - esperado_m

        if row["gols_mandante"] > row["gols_visitante"]:
            resultado_m = 1
            resultado_v = 0

        elif row["gols_mandante"] < row["gols_visitante"]:
            resultado_m = 0
            resultado_v = 1

        else:
            resultado_m = 0.5
            resultado_v = 0.5

        elos[mandante] = elo_m + K * (resultado_m - esperado_m)
        elos[visitante] = elo_v + K * (resultado_v - esperado_v)

        ano_anterior = ano_atual

    return elos
```

```python
def montar_features_jogo(
    mandante,
    visitante,
    estado_simulacao,
    estado_mando_simulacao
):

    # Estado geral dos dois times
    dados_mandante = estado_simulacao[
        estado_simulacao["time"] == mandante
    ].iloc[0]

    dados_visitante = estado_simulacao[
        estado_simulacao["time"] == visitante
    ].iloc[0]


    # Estado específico de mando
    # mandante joga em casa -> casa = 1
    mando_mandante = estado_mando_simulacao[
        (estado_mando_simulacao["time"] == mandante) &
        (estado_mando_simulacao["casa"] == 1)
    ].iloc[0]

    # visitante joga fora -> casa = 0
    mando_visitante = estado_mando_simulacao[
        (estado_mando_simulacao["time"] == visitante) &
        (estado_mando_simulacao["casa"] == 0)
    ].iloc[0]


    # Montar features
    features = {

        # ELO
        "elo_mandante_sazonal":
            dados_mandante["elo_atual"],

        "elo_visitante_sazonal":
            dados_visitante["elo_atual"],

        "elo_sazonal_diff":
            dados_mandante["elo_atual"]
            - dados_visitante["elo_atual"],


        # EWM geral mandante
        "pontos_ewm_mandante":
            dados_mandante["pontos_ewm_atual"],

        "gols_ewm_mandante":
            dados_mandante["gols_ewm_atual"],

        "gols_sofridos_ewm_mandante":
            dados_mandante["gols_sofridos_ewm_atual"],

        "vitoria_ewm_mandante":
            dados_mandante["vitoria_ewm_atual"],


        # EWM geral visitante
        "pontos_ewm_visitante":
            dados_visitante["pontos_ewm_atual"],

        "gols_ewm_visitante":
            dados_visitante["gols_ewm_atual"],

        "gols_sofridos_ewm_visitante":
            dados_visitante["gols_sofridos_ewm_atual"],

        "vitoria_ewm_visitante":
            dados_visitante["vitoria_ewm_atual"],


        # EWM mando mandante
        "pontos_ewm_mando_mandante":
            mando_mandante["pontos_ewm_mando_atual"],

        "gols_ewm_mando_mandante":
            mando_mandante["gols_ewm_mando_atual"],

        "gols_sofridos_ewm_mando_mandante":
            mando_mandante["gols_sofridos_ewm_mando_atual"],

        "vitoria_ewm_mando_mandante":
            mando_mandante["vitoria_ewm_mando_atual"],


        # EWM mando visitante
        "pontos_ewm_mando_visitante":
            mando_visitante["pontos_ewm_mando_atual"],

        "gols_ewm_mando_visitante":
            mando_visitante["gols_ewm_mando_atual"],

        "gols_sofridos_ewm_mando_visitante":
            mando_visitante["gols_sofridos_ewm_mando_atual"],

        "vitoria_ewm_mando_visitante":
            mando_visitante["vitoria_ewm_mando_atual"],
    }

    # Diferenças gerais
    features["pontos_ewm_diff"] = (
        features["pontos_ewm_mandante"]
        - features["pontos_ewm_visitante"]
    )

    features["gols_ewm_diff"] = (
        features["gols_ewm_mandante"]
        - features["gols_ewm_visitante"]
    )

    features["gols_sofridos_ewm_diff"] = (
        features["gols_sofridos_ewm_mandante"]
        - features["gols_sofridos_ewm_visitante"]
    )

    features["vitoria_ewm_diff"] = (
        features["vitoria_ewm_mandante"]
        - features["vitoria_ewm_visitante"]
    )


    # Diferenças por mando
    features["pontos_ewm_mando_diff"] = (
        features["pontos_ewm_mando_mandante"]
        - features["pontos_ewm_mando_visitante"]
    )

    features["gols_ewm_mando_diff"] = (
        features["gols_ewm_mando_mandante"]
        - features["gols_ewm_mando_visitante"]
    )

    features["gols_sofridos_ewm_mando_diff"] = (
        features["gols_sofridos_ewm_mando_mandante"]
        - features["gols_sofridos_ewm_mando_visitante"]
    )

    features["vitoria_ewm_mando_diff"] = (
        features["vitoria_ewm_mando_mandante"]
        - features["vitoria_ewm_mando_visitante"]
    )


    # vira DataFrame com 1 linha
    X_jogo = pd.DataFrame(
        [features],
        columns=features_modelo_B
    )

    return X_jogo
```

```python
def calcular_lambdas(
    mandante,
    visitante,
    estado_simulacao,
    estado_mando_simulacao
):

    # dados gerais
    dados_m = estado_simulacao[
        estado_simulacao["time"] == mandante
    ].iloc[0]

    dados_v = estado_simulacao[
        estado_simulacao["time"] == visitante
    ].iloc[0]

    # dados por mando
    mando_m = estado_mando_simulacao[
        (estado_mando_simulacao["time"] == mandante) &
        (estado_mando_simulacao["casa"] == 1)
    ].iloc[0]

    mando_v = estado_mando_simulacao[
        (estado_mando_simulacao["time"] == visitante) &
        (estado_mando_simulacao["casa"] == 0)
    ].iloc[0]

    # ataque recente
    ataque_m = (
        dados_m["gols_ewm_atual"]
        + mando_m["gols_ewm_mando_atual"]
    ) / 2

    ataque_v = (
        dados_v["gols_ewm_atual"]
        + mando_v["gols_ewm_mando_atual"]
    ) / 2

    # defesa adversária
    defesa_v = (
        dados_v["gols_sofridos_ewm_atual"]
        + mando_v["gols_sofridos_ewm_mando_atual"]
    ) / 2

    defesa_m = (
        dados_m["gols_sofridos_ewm_atual"]
        + mando_m["gols_sofridos_ewm_mando_atual"]
    ) / 2

    # gols esperados
    lambda_m = (ataque_m + defesa_v) / 2
    lambda_v = (ataque_v + defesa_m) / 2

    return lambda_m, lambda_v
```

```python
def sortear_placar(resultado, lambda_m, lambda_v, rng):

    while True:

        gols_m = rng.poisson(lambda_m)
        gols_v = rng.poisson(lambda_v)

        # vitória mandante
        if resultado == "H" and gols_m > gols_v:
            return gols_m, gols_v

        # empate
        if resultado == "D" and gols_m == gols_v:
            return gols_m, gols_v

        # vitória visitante
        if resultado == "A" and gols_m < gols_v:
            return gols_m, gols_v
```

```python
def atualizar_estado_jogo(
    mandante,
    visitante,
    gols_m,
    gols_v,
    estado_simulacao,
    estado_mando_simulacao,
    span=5,
    K=18
):
    # alpha usado pela EWM(span=5, adjust=False)
    alpha = 2 / (span + 1)

    # localizar os dois times no estado geral
    idx_m = estado_simulacao[
        estado_simulacao["time"] == mandante
    ].index[0]

    idx_v = estado_simulacao[
        estado_simulacao["time"] == visitante
    ].index[0]

    # --------------------------------
    # 1. Resultado: pontos e vitória
    # --------------------------------

    if gols_m > gols_v:
        pontos_m, pontos_v = 3, 0
        vitoria_m, vitoria_v = 1, 0

        resultado_elo_m = 1
        resultado_elo_v = 0

    elif gols_m < gols_v:
        pontos_m, pontos_v = 0, 3
        vitoria_m, vitoria_v = 0, 1

        resultado_elo_m = 0
        resultado_elo_v = 1

    else:
        pontos_m, pontos_v = 1, 1
        vitoria_m, vitoria_v = 0, 0

        resultado_elo_m = 0.5
        resultado_elo_v = 0.5


    # --------------------------------
    # 2. Atualizar classificação
    # --------------------------------

    estado_simulacao.at[idx_m, "jogos"] += 1
    estado_simulacao.at[idx_v, "jogos"] += 1

    estado_simulacao.at[idx_m, "pontos"] += pontos_m
    estado_simulacao.at[idx_v, "pontos"] += pontos_v

    estado_simulacao.at[idx_m, "vitorias"] += vitoria_m
    estado_simulacao.at[idx_v, "vitorias"] += vitoria_v

    estado_simulacao.at[idx_m, "gols_pro"] += gols_m
    estado_simulacao.at[idx_m, "gols_contra"] += gols_v

    estado_simulacao.at[idx_v, "gols_pro"] += gols_v
    estado_simulacao.at[idx_v, "gols_contra"] += gols_m

    estado_simulacao.at[idx_m, "saldo_gols"] = (
        estado_simulacao.at[idx_m, "gols_pro"]
        - estado_simulacao.at[idx_m, "gols_contra"]
    )

    estado_simulacao.at[idx_v, "saldo_gols"] = (
        estado_simulacao.at[idx_v, "gols_pro"]
        - estado_simulacao.at[idx_v, "gols_contra"]
    )


    # --------------------------------
    # 3. Atualizar EWM geral
    # nova_ewm = alpha * valor_novo
    #          + (1-alpha) * ewm_anterior
    # --------------------------------

    def atualizar_ewm(valor_antigo, valor_novo):
        return (
            alpha * valor_novo
            + (1 - alpha) * valor_antigo
        )

    # mandante
    estado_simulacao.at[idx_m, "pontos_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_m, "pontos_ewm_atual"],
        pontos_m
    )

    estado_simulacao.at[idx_m, "gols_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_m, "gols_ewm_atual"],
        gols_m
    )

    estado_simulacao.at[idx_m, "gols_sofridos_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_m, "gols_sofridos_ewm_atual"],
        gols_v
    )

    estado_simulacao.at[idx_m, "vitoria_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_m, "vitoria_ewm_atual"],
        vitoria_m
    )

    # visitante
    estado_simulacao.at[idx_v, "pontos_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_v, "pontos_ewm_atual"],
        pontos_v
    )

    estado_simulacao.at[idx_v, "gols_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_v, "gols_ewm_atual"],
        gols_v
    )

    estado_simulacao.at[idx_v, "gols_sofridos_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_v, "gols_sofridos_ewm_atual"],
        gols_m
    )

    estado_simulacao.at[idx_v, "vitoria_ewm_atual"] = atualizar_ewm(
        estado_simulacao.at[idx_v, "vitoria_ewm_atual"],
        vitoria_v
    )


    # --------------------------------
    # 4. Atualizar EWM por mando
    # --------------------------------

    idx_mando_m = estado_mando_simulacao[
        (estado_mando_simulacao["time"] == mandante) &
        (estado_mando_simulacao["casa"] == 1)
    ].index[0]

    idx_mando_v = estado_mando_simulacao[
        (estado_mando_simulacao["time"] == visitante) &
        (estado_mando_simulacao["casa"] == 0)
    ].index[0]

    # mandante em casa
    estado_mando_simulacao.at[
        idx_mando_m, "pontos_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_m, "pontos_ewm_mando_atual"
        ],
        pontos_m
    )

    estado_mando_simulacao.at[
        idx_mando_m, "gols_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_m, "gols_ewm_mando_atual"
        ],
        gols_m
    )

    estado_mando_simulacao.at[
        idx_mando_m, "gols_sofridos_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_m, "gols_sofridos_ewm_mando_atual"
        ],
        gols_v
    )

    estado_mando_simulacao.at[
        idx_mando_m, "vitoria_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_m, "vitoria_ewm_mando_atual"
        ],
        vitoria_m
    )

    # visitante fora
    estado_mando_simulacao.at[
        idx_mando_v, "pontos_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_v, "pontos_ewm_mando_atual"
        ],
        pontos_v
    )

    estado_mando_simulacao.at[
        idx_mando_v, "gols_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_v, "gols_ewm_mando_atual"
        ],
        gols_v
    )

    estado_mando_simulacao.at[
        idx_mando_v, "gols_sofridos_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_v, "gols_sofridos_ewm_mando_atual"
        ],
        gols_m
    )

    estado_mando_simulacao.at[
        idx_mando_v, "vitoria_ewm_mando_atual"
    ] = atualizar_ewm(
        estado_mando_simulacao.at[
            idx_mando_v, "vitoria_ewm_mando_atual"
        ],
        vitoria_v
    )


    # --------------------------------
    # 5. Atualizar Elo
    # --------------------------------

    elo_m = estado_simulacao.at[idx_m, "elo_atual"]
    elo_v = estado_simulacao.at[idx_v, "elo_atual"]

    esperado_m = 1 / (
        1 + 10 ** ((elo_v - elo_m) / 400)
    )

    esperado_v = 1 - esperado_m

    novo_elo_m = (
        elo_m
        + K * (resultado_elo_m - esperado_m)
    )

    novo_elo_v = (
        elo_v
        + K * (resultado_elo_v - esperado_v)
    )

    estado_simulacao.at[idx_m, "elo_atual"] = novo_elo_m
    estado_simulacao.at[idx_v, "elo_atual"] = novo_elo_v

    return estado_simulacao, estado_mando_simulacao
```

```python
def simular_temporada(
    jogos_restantes,
    estado_inicial,
    estado_mando_inicial,
    modelo,
    rng
):

    # Cada temporada começa da realidade atual
    estado_simulacao = estado_inicial.copy()
    estado_mando_simulacao = estado_mando_inicial.copy()

    resultados_jogos = []

    # ordem dos jogos
    jogos = jogos_restantes.sort_values(
        ["data", "rodada"]
    ).copy()

    for _, jogo in jogos.iterrows():

        mandante = jogo["time_mandante"]
        visitante = jogo["time_visitante"]

        # 1. Montar as 27 features NO ESTADO ATUAL
        X_jogo = montar_features_jogo(
            mandante,
            visitante,
            estado_simulacao,
            estado_mando_simulacao
        )

        # 2. Probabilidades do modelo
        proba = modelo.predict_proba(X_jogo)[0]

        classes = modelo.named_steps[
            "logistica"
        ].classes_

        # 3. Sortear H / D / A
        resultado = sortear_resultado(
            classes,
            proba,
            rng
        )

        # 4. Calcular médias esperadas de gols
        lambda_m, lambda_v = calcular_lambdas(
            mandante,
            visitante,
            estado_simulacao,
            estado_mando_simulacao
        )

        # 5. Sortear placar compatível com H/D/A
        gols_m, gols_v = sortear_placar(
            resultado,
            lambda_m,
            lambda_v,
            rng
        )

        # 6. Guardar resultado
        resultados_jogos.append({
            "rodada": jogo["rodada"],
            "mandante": mandante,
            "visitante": visitante,
            "gols_mandante": gols_m,
            "gols_visitante": gols_v,
            "resultado": resultado
        })

        # 7. Atualizar o mundo simulado
        estado_simulacao, estado_mando_simulacao = (
            atualizar_estado_jogo(
                mandante,
                visitante,
                gols_m,
                gols_v,
                estado_simulacao,
                estado_mando_simulacao
            )
        )

    # classificação ao fim da temporada
    classificacao = estado_simulacao.sort_values(
        [
            "pontos",
            "vitorias",
            "saldo_gols",
            "gols_pro"
        ],
        ascending=False
    ).reset_index(drop=True)

    classificacao["posicao"] = (
        classificacao.index + 1
    )

    resultados_jogos = pd.DataFrame(
        resultados_jogos
    )

    return classificacao, resultados_jogos
```

mostrar as porcentagenms de cada time ficar em determinada posição na tabela

```python
# Probabilidade de cada time terminar em cada posição
prob_posicao = pd.crosstab(
    df_mc["time"],
    df_mc["posicao"]
)

prob_posicao = prob_posicao.reindex(
    columns=range(1, 21),
    fill_value=0
)

prob_posicao = (
    prob_posicao
    .div(n_simulacoes)
    .mul(100)
)
```

```python
from scipy.optimize import linear_sum_assignment

matriz = prob_posicao.to_numpy()

# negativo porque a função minimiza;
# queremos maximizar as probabilidades
linhas, colunas = linear_sum_assignment(-matriz)

tabela_projetada = pd.DataFrame({
    "time": prob_posicao.index[linhas],
    "posicao": prob_posicao.columns[colunas],
    "probabilidade_%": matriz[linhas, colunas]
})

tabela_projetada = (
    tabela_projetada
    .sort_values("posicao")
    .reset_index(drop=True)
)

tabela_projetada["probabilidade_%"] = (
    tabela_projetada["probabilidade_%"]
    .round(1)
)

tabela_projetada
```
